```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("sales_data_sample.csv",encoding='latin1')
```

```python
df['ADDRESSLINE2'].fillna(df['ADDRESSLINE2'].mode()[0])
```

```python
df['ADDRESSLINE2'].mode()
```

```python
df['ADDRESSLINE2'] = df['ADDRESSLINE2'].fillna('Unknown')
```

```python
df['STATE'] = df['STATE'].fillna(df['STATE'].mode()[0])
```

```python
df['POSTALCODE'] = df['POSTALCODE'].fillna(df['POSTALCODE'].mode()[0])
```

```python
df['TERRITORY'] = df['TERRITORY'].fillna(df['TERRITORY'].mode()[0])
```

```python
df.duplicated().sum()
```

```python
df["YEAR_ID"] = pd.to_datetime(df["YEAR_ID"])
```

```python
print(df["MONTH_ID"].dtype)
```

```python
print(df["MONTH_ID"].unique())
```

```python
total_sales = df["SALES"].sum()
print(total_sales)
```

```python
average_sales = df["SALES"].mean()
print(average_sales)
```

```python
SALES_STATUS = (
    df.groupby("STATUS")["SALES"]
    .sum()
    .sort_values(ascending=False)
)

print(SALES_STATUS)
```

```python
STATE_SALES = (
    df.groupby("STATE")["SALES"]
    .sum()
    .sort_values(ascending=False)
)

print(STATE_SALES)
```

```python
df["ORDERDATE"] = pd.to_datetime(df["ORDERDATE"])
```

```python
df["MONTH_ID"] = df["ORDERDATE"].dt.to_period("M")
monthly_sales = df.groupby("MONTH_ID")["SALES"].sum()
monthly_sales.plot(kind="line", figsize=(10, 5))
plt.title("Monthly Sales Trend")
plt.xlabel("MONTH_ID")
plt.ylabel("SALES")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["SALES"], bins=20)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df["SALES"])

plt.title("Sales Distribution and Outliers")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["QUANTITYORDERED"],
    df["SALES"],
    alpha=0.5
)

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```
